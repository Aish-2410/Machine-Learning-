```python
import pandas as pd
data=[ 
    ["Sunny","Warm","Normal","Strong","Warm","Same","Yes"],
     ["Sunny","Warm","High","Strong","Warm","Same","Yes"],
     ["Rainy","Cold","High","Strong","Warm","Change","No"],
     ["Sunny","Warm","High","Strong","Cool","Change","Yes"]
]

attributes = ["Sky","Airtemp","Humidity","Wind","Water","Forecast","EnjoySport"]

print("The Dataset is: ")

df = pd.DataFrame(data, columns=attributes)

df
    
```

```python
def finds(data):
    h = ["0"] * (len(data[0])-1)
    for row in data:
        features = row[:-1]
        target = row[-1]
        
        if target=="Yes":
            for i in range(len(h)):
                if h[i]=="0":
                    h[i] = features[i]
                    
                if h[i]!=features[i]:
                    h[i] = "?"
    return h
                                         
finds(data)
    
```

```python
#validation
def valid(data):
    final = finds(data)
    for row in data:
            features = row[:-1]
            target = row[-1]

            if target=="Yes":
                for i in range(len(final)):
                    if(final[i]==data[0][i] or final[i]=='?'):
                        valid = True;
                    else:
                        valid = False;

    return valid

```

```python

data2 = [["Red","Hard","Yes","Smooth","Yes"],
        ["Brown","Soft","No","Wrinkled","No"],
        ["Red","Hard","Yes","Scaly","Yes"],
        ["Yellow","Hard","Yes","Smooth","Yes"]]

attr = ["Color","Toughness","Fungus","Appearance","Poisonous"]
df2 = pd.DataFrame(data2, columns = attr)

df2

final = finds(data2)
isvalid = valid(data2)

print(final)
if(isvalid):
    print("The hypothesis is correct")
else:
    print("the hypothesis is incorrect")
```

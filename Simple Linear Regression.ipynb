**SIMPLE LINEAR REGRESSION**

```python
#importing libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
```

```python
#loading salary dataset
df = pd.read_csv("Salary_Data.csv")
df
```

```python
x = df['YearsExperience']
y = df['Salary']

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state=12)

print("X_train: \n")
print(x_train)
print("\nX_test: \n")
print(x_test)
print("\nY_train: \n")
print(y_train)
print("\nY_test: \n")
print(y_test)
```

```python
x_train_matrix = x_train.values.reshape(-1,1)
reg = LinearRegression()
reg.fit(x_train_matrix,y_train)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.score(x_train_matrix,y_train)*100
```

```python
x_test_matrix = x_test.values.reshape(-1,1)
y_hat_train = reg.predict(x_train_matrix)
y_hat_test = reg.predict(x_test_matrix)

plt.scatter(x_train,y_train,color='blue')
fig = plt.plot(x_train,y_hat_train,color='red',label='Linear Regression')
```

```python
x_test_matrix = x_test.values.reshape(-1,1)
y_hat_test = reg.predict(x_test_matrix)

plt.scatter(x_test,y_test,color='blue')
fig = plt.plot(x_test,y_hat_test,color='red',label='Linear Regression')
```

```python
testx = [1.2,1.3,1.5,3,8,3,2,7]

df2 = pd.DataFrame(testx,columns=['Years_test'])
df2['Predicted Salary'] = reg.predict(df2)
df2
```

```python
mae = mean_absolute_error(y_test,y_hat_test)
mse = mean_squared_error(y_test,y_hat_test)
r2 = r2_score(y_test,y_hat_test)

print("MAE: ",mae)
print("MSE: ",mse)
print("R2: ",r2)
```
